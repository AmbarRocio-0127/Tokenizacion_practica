#### *Importación de Librerías*

```python
import nltk
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from collections import Counter
import string
```

```python
def procesar_texto_pln(texto, idioma='spanish'): 
    print('\n☑️Inicializando El Programa PLN☑️')
    
    print('\nPaso 1 : Tokenizacion de texto✔️')
    try:
        tokens = word_tokenize(texto.lower(), language=idioma)
        print('✅¡Tokenización Inicializada! ✅')
    except LookupError:
        print("❌ Error: recursos de NLTK no encontrados. Ejecuta nltk.download('punkt_tab').")
        return None
    
    print('\n🧹Paso 2: Limpieza de Tokens (stopwords y puntuación) 🗑️💻')
    stop_words = set(stopwords.words(idioma))
    
    tokens_limpios = [
        token for token in tokens
        if token.isalnum() and token not in stop_words
    ]
    
    print('\n🔢 Paso 3: Calcular la Frecuencia, Contar las palabras🔍')
    
    frecuencia_palabras = Counter(tokens_limpios)
    
    palabras_comunes = frecuencia_palabras.most_common(10)
    
    return{
        "tokens_originales": tokens,
        "tokens_limpios": tokens_limpios, 
        "frecuencias": dict(frecuencia_palabras),
        "top_10": palabras_comunes
    }
    
if __name__ == "__main__":
    
        texto_ejemplo = """El Alquimista de Paulo Coelho es una célebre novela alegórica sobre Santiago, un joven pastor andaluz que 
        emprende un viaje físico y espiritual desde España hasta las Pirámides de Egipto tras un sueño recurrente. 
        Guiado por figuras como Melquisedec y un poderoso alquimista, Santiago aprende a escuchar a su corazón, 
        descubrir su "Leyenda Personal" y comprender el "Lenguaje del Mundo". A lo largo de su travesía por el desierto 
        del Sahara, experimenta pérdidas, trabajo arduo y el amor verdadero junto a Fátima. Finalmente, tras superar 
        múltiples pruebas, comprende que la verdadera riqueza de su travesía radica en la sabiduría y la transformación 
        interior alcanzadas en el camino.

        En su travesía, Santiago aprende del Alquimista a escuchar su corazón y supera el miedo al sufrimiento. En el 
        oasis de Al-Fayum encuentra el amor en Fátima, quien fortalece su búsqueda de la Leyenda Personal. Finalmente, 
        al llegar a las Pirámides, unos bandidos lo asaltan, y el líder le revela sin querer que el tesoro siempre estuvo 
        en España, bajo el sicomoro de la iglesia donde dormía con sus ovejas.
        """
    
        print("\n📥Texto de entrada 📥")
        print(texto_ejemplo.strip())
    
        resultados = procesar_texto_pln(texto_ejemplo)
    
        if resultados:
            print("\n 🎯 Resultados Finales 📈")
            print(f"\n 🔢 Top 10 palabras mas relevantes (Frecuencia) 🔝")
            for palabra, frecuencia in resultados["top_10"]:
                print(f"- '{palabra}': aparece {frecuencia} vez/veces")
            
            print(f"\n🔤📚Total de palabras únicas (Vocabulario): {len(resultados['frecuencias'])} ")
        
        else:
            print("\n🚨Ocurrió un error en el procesamiento.🚫")
```
